## Regular Expressions

In Python, regular expressions are a powerful tool for searching and manipulating strings based on specific patterns. The `re` module provides functions to work with regular expressions.

```python
import re
```

### Some Commonly Used Functions in the `re` Module:

- `re.match()`: Checks for a match only at the beginning of the string.
- `re.search()`: Searches the entire string for a match.
- `re.findall()`: Returns a list of all matches in the string.
- `re.sub()`: Replaces occurrences of a pattern with a specified replacement string.
- `re.split()`: Splits the string by the occurrences of the pattern.
- `re.group()`: Returns the string matched by a specific group in a match object.

Special parameters in the `re` module include:

- `re.MULTILINE`: Changes the behavior of `^` and `$` to match the start and end of each line, respectively, rather than the start and end of the entire string.
- `re.IGNORECASE`: Makes the pattern matching case-insensitive.
- `re.DOTALL`: Allows the `.` character to match newline characters as well, enabling the pattern to match across multiple lines.

In Natural Language Processing (NLP), regular expressions are often used for tasks such as tokenization, text cleaning, and pattern matching. They can help identify specific patterns in text, such as email addresses, phone numbers, or specific keywords.

Some key expressions include in table

| Expression | Description |
|------------|-------------|
| `\d`       | Matches any digit (0-9) |
| `\w`       | Matches any alphanumeric character (letters and digits) |
| `\s`       | Matches any whitespace character (spaces, tabs, etc.) |
| `.`        | Matches any character except a newline |
| `^`        | Matches the start of a string |
| `$`        | Matches the end of a string |    
| `*`        | Matches 0 or more repetitions of the preceding pattern |
| `+`        | Matches 1 or more repetitions of the preceding pattern |
| `?`        | Matches 0 or 1 repetition of the preceding pattern |
| `!`        | Negative lookahead assertion |
| `{n}`      | Matches exactly n repetitions of the preceding pattern |
| `{n,}`     | Matches n or more repetitions of the preceding pattern |
| `{n,m}`    | Matches between n and m repetitions of the preceding pattern |
| `[]`       | Matches any one of the characters inside the brackets |
| `\|`        | Acts as a logical OR between patterns |
| `()`       | Groups patterns together and captures the matched text |
| `\b`       | Matches a word boundary (the position between a word character and a non-word character) |
| `@ `        | Matches the literal "@" character |
| `\1 `        | Refers to the text captured by the first group |

Regular expressions can be combined to create complex patterns for matching specific text structures. For example, to match an email address, you might use a pattern like `\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b`, which checks for a sequence of characters that resembles an email format.

#### Some commonly used regular expression patterns are:

- **Disjunctions**: Disjunctions are used to match one of several possible patterns. In regular expressions, the `|` operator is used to denote a disjunction. For example, the pattern `cat|dog` will match either "cat" or "dog" in a given string.

- **Negation in Disjunctions**: Negation in disjunctions allows you to specify patterns that should not be matched. In regular expressions, the `^` operator can be used inside a character class to indicate negation. For example, the pattern `[^aeiou]` will match any character that is not a vowel.

- **Kleene Star  and Kleene Plus**: The Kleene star (`*`) and Kleene plus (`+`) are quantifiers that specify how many times a pattern should be matched. The Kleene star matches zero or more occurrences of the preceding pattern, while the Kleene plus matches one or more occurrences. For example, the pattern `a*` will match an empty string, "a", "aa", "aaa", and so on, while the pattern `a+` will match "a", "aa", "aaa", but not an empty string.

- **Grouping and Capturing**: Grouping allows you to treat multiple characters as a single unit in a regular expression. This is done using parentheses `()`. Capturing groups also allow you to extract the matched text for further processing. For example, the pattern `(cat|dog)` will match either "cat" or "dog", and you can access the matched text using the captured group.

- **Character Classes**: Character classes are used to match any one of a set of characters. They are defined by enclosing the characters in square brackets. For example, the pattern `[aeiou]` will match any vowel.

- **Quantifiers**: Quantifiers are used to specify how many times a character or group should be matched. The most common quantifiers are `*` (zero or more), `+` (one or more), and `?` (zero or one).

- **Anchors**: Anchors are used to match the position of a character in a string rather than the character itself. The `^` anchor matches the start of a string, while the `$` anchor matches the end of a string.

- **Wildcards**: The `.` character is a wildcard that matches any single character except a newline. For example, the pattern `a.c` will match "abc", "a1c", "a-c", etc.

We are given this passage:

```python
passage = """
Humbert stepped cautiously into the damp grotto, his flashlight cutting through the heavy darkness.
High above, a startled raven croaked and flapped its wings against the stone ceiling.
On the cave floor, Humbert froze as he spotted a strange insect scurrying near his boot.
He initially thought it was just an ordinary beetle, but as he moved closer, he realized the bug was much larger than expected.
It was, in fact, the big bug the locals had warned him about.
222 Sandora echoed through the cave as an ancient spell was activated.
"Humbert! Humbert!" a panicked voice echoed from the cave entrance.
"""
```

```python
# All Unique Strings as a List
words = re.findall(r"\b[a-zA-Z]+\b", passage)

unique_words = sorted(set(words))
print(unique_words)
```

```python
# All Unique Strings that End with t
matches = re.findall(r"\b[a-zA-Z]*t\b", passage)

unique_matches = sorted(set(matches))
print(unique_matches)
```

```python
# Strings that Start with b and End with g
# Where the middle character is a vowel (a e i o u)
matches = re.findall(r"\bb[aeiou]+g\b", passage)
print(matches)
```

```python
# Consecutive Repeated Words
matches = re.findall(r"\b(\w+)\W+\1\b", passage)
print(matches)
```

```python
# Strings that Start with an Integer and End with a Word
matches = re.findall(r"^\d+.*?[a-zA-Z]+\b\.?$", passage, re.MULTILINE)
print(matches)
```

```python
# Strings Containing Both "grotto" and "raven"
match = re.search(r"\bgrotto\b.*\braven\b|\braven\b.*\bgrotto\b", passage, re.DOTALL)
print(bool(match))
```

```python
# Extract the Part Starting with "flashlight" and Ending with "darkness"
match = re.search(r"flashlight.*?darkness", passage)
print(match.group())
```

```python
# Extract the First Word from the Sentence
sentence = re.search(r"flashlight.*?darkness", passage)
first_word = re.match(r"\b\w+\b", sentence.group())
print(first_word.group())
```

```python
# Contain grotto or cave floor
matches = re.findall(r'\b(grotto|cave floor)\b', passage)
print(matches)
```

### Regex Substitutions

Regular expressions can also be used for substitutions, allowing you to replace specific patterns in a string with new text. The `re.sub()` function is commonly used for this purpose. It takes three main arguments: the pattern to search for, the replacement string, and the original string.

```python
# Basic Regex Substitutions
text = "I like cherry pie."

result = re.sub(r"cherry", r"apricot", text)
print(result)
```

```python
# Converting Text Format
text = "janet is here."
result = re.sub(r"janet", r"Janet", text)
print(result)
```

**Converting Date Formats**: Suppose we want to convert US format `mm/dd/yyyy` to European format `dd/mm/yyyy`. We can use the following code:

Groups are used to capture specific parts of a match for later use. They are defined using parentheses `()`. For example, in the pattern `(\d{2})/(\d{2})/(\d{4})`, we have three groups that capture the month, day, and year from a date string. We can then refer to these groups in the replacement string using `\1`, `\2`, and `\3` respectively.

```python
# Step 1 — Create Groups
# Pattern: r"(\d{2})/(\d{2})/(\d{4})". This creates three capture groups:
# Group 1: (\d{2}) captures the month (mm).
# Group 2: (\d{2}) captures the day (dd).
# Group 3: (\d{4}) captures the year (yyyy).

# Step 2 — Rearrange Groups
# Replacement: r"\2/\1/\3". This rearranges the captured groups to the desired format:
# \2 refers to the second capture group (day).
# \1 refers to the first capture group (month).
# \3 refers to the third capture group (year).

text = "Today's date is 14/02/2026"

result = re.sub(r"(\d{2})/(\d{2})/(\d{4})", r"\2-\1-\3", text)
print(result)
```

**Non-Capturing Groups**: Non-capturing groups are used when you want to group part of a regular expression but do not need to capture the matched text for later use. They are defined using `(?:...)`. For example, the pattern `(?:cat|dog)` will match either "cat" or "dog", but it will not create a capturing group for the matched text.

```python
pattern = r"(?:some|a few) (people|cats) like some \1"

text1 = "some cats like some cats"
text2 = "some cats like some some"

print(bool(re.search(pattern, text1)))
print(bool(re.search(pattern, text2)))
```

**Selective Capturing**: Sometimes, you may want to capture only specific parts of a match while ignoring others. This can be achieved by using non-capturing groups for the parts you want to ignore and regular capturing groups for the parts you want to capture. For example, the pattern `(?:\d{2})/(\d{2})/(?:\d{4})` will only capture the day from a date string in the format `mm/dd/yyyy`, while ignoring the month and year.

```python
# This one means among 14 occurrences of the date format, we want to capture the 15th one. 
# The non-capturing group (?:\d\d/\d\d/\d\d\d\d\s+) matches the first 14 dates,
# and the capturing group (\d\d/\d\d/\d\d\d\d) captures the 15th date.
text = ("01/01/2026 02/01/2026 03/01/2026 04/01/2026 " 
        "05/01/2026 06/01/2026 07/01/2026 08/01/2026 " 
        "09/01/2026 10/01/2026 11/01/2026 12/01/2026 " 
        "13/01/2026 14/01/2026 15/01/2026")

match = re.search(r"(?:\d\d/\d\d/\d\d\d\d\s+){14}(\d\d/\d\d/\d\d\d\d)", text)
print(match.group(1))
```

**Lookahead Assertions**: Lookahead assertions are used to match a pattern only if it is followed by another specific pattern. They are defined using `(?=...)`. For example, the pattern `\d+(?= dollars)` will match a sequence of digits only if it is followed by the word "dollars". This allows you to find numbers that are specifically associated with a certain context without including the context in the match itself.

```python
# Example — First Word Not Starting with T
text1 = "Hello world"
text2 = "Today is sunny"

print(re.search(r"^(?![tT])(\w+)\b", text1))
print(re.search(r"^(?![tT])(\w+)\b", text2))
```

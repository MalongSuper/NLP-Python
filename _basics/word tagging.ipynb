### POS Tagging

Part-of-speech (POS) tagging is the process of assigning a part of speech to each word in a sentence, such as noun, verb, adjective, etc. This is an important step in natural language processing (NLP) as it helps in understanding the grammatical structure of sentences and can be used for various applications like parsing, information extraction, and machine translation.

For example: 'Adam is currently working on a project. I think he is doing a great job.' 
could be tagged as:
- Adam (NNP - Proper Noun)
- is (VBZ - Verb, 3rd person singular present)
- currently (RB - Adverb)
- working (VBG - Verb, gerund or present participle)
- on (IN - Preposition)
- a (DT - Determiner)
- project (NN - Noun, singular)
- I (PRP - Personal Pronoun)
- think (VBP - Verb, non-3rd person singular present)
- he (PRP - Personal Pronoun)
- is (VBZ - Verb, 3rd person singular present)
- doing (VBG - Verb, gerund or present participle)

````python
import nltk

nltk.download('stopwords')
nltk.download('punkt_tab') 
nltk.download('averaged_perceptron_tagger_eng')
````

````python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize

stop_words = set(stopwords.words('english'))
text = 'Adam is currently working on a project. I think he is doing a great job.' 

tokenized = sent_tokenize(text)

for sentence in tokenized:
    words = word_tokenize(sentence)
    words = [w for w in words if w.lower() not in stop_words]
    tagged = nltk.pos_tag(words)
    print(tagged)
````

#### POS Tagging with Hidden Markov Models (HMM)

In POS tagging, Hidden Markov Models (HMM) are often used to model the sequence of tags in a sentence. HMMs are probabilistic models that assume the current state (tag) depends only on the previous state (tag) and the current observation (word).
For example: We have a statement: "He will..."
- He (PRP - Personal Pronoun)
- will (MD - Modal Verb)

In this case, the HMM would model the probability of "will" being a modal verb given that "He" is a personal pronoun. The model would also consider the likelihood of "He" being a personal pronoun based on the context of the sentence.
Mathematically, the HMM can be represented as:

$P(T|W) = P(T) * P(W|T)$

Where:
- $P(T|W)$ is the probability of a sequence of tags T given a sequence of words W.
- $P(T)$ is the probability of the sequence of tags T.
- $P(W|T)$ is the probability of the sequence of words W given the sequence of tags T.

In this case, the probability of "will" being a modal verb given that "He" is a personal pronoun can be calculated using the HMM model, which would help in accurately tagging the parts of speech in the sentence. Mathematically, we can express this as:

$P(MD|PRP) = P(MD) * P(PRP|MD)$

Where:
- $P(MD|PRP)$ is the probability of "will" being a modal verb given that "He" is a personal pronoun.

- $P(MD)$ is the probability of "will" being a modal verb in general.

- $P(PRP|MD)$ is the probability of "He" being a personal pronoun given that "will" is a modal verb.

To train an HMM for POS tagging, you would typically use a labeled corpus of text where each word is tagged with its corresponding part of speech. The training process involves estimating the probabilities of tag sequences and word-tag associations based on the frequency of occurrences in the training data.


Reference: https://www.geeksforgeeks.org/nlp/part-of-speech-pos-tagging-with-hidden-markov-model/

````python
train_data = [
    [("I", "PRON"), ("eat", "VERB"), ("rice", "NOUN")],
    [("You", "PRON"), ("eat", "VERB"), ("fish", "NOUN")],
    [("They", "PRON"), ("like", "VERB"), ("rice", "NOUN")],
    [("We", "PRON"), ("like", "VERB"), ("fish", "NOUN")],
    [("Cats", "NOUN"), ("eat", "VERB"), ("fish", "NOUN")]
]
````

Transition and observation probabilities are estimated from the training data. 

- The transition probabilities (denoted as $A$) represent the likelihood of a tag following another tag in a sequence. For example, the probability of a noun being followed by a verb.

- The observation probabilities (denoted as $B$) represent the likelihood of a word being associated with a particular tag. For example, the probability of the word "run" being tagged as a verb versus a noun.

````python
from collections import defaultdict
import math

transition = defaultdict(lambda: defaultdict(int))
emission = defaultdict(lambda: defaultdict(int))
start_prob = defaultdict(int)
tag_counts = defaultdict(int)

for sentence in train_data:
    print(sentence)
    prev_tag = None
    for i, (word, tag) in enumerate(sentence):
        tag_counts[tag] += 1
        emission[tag][word] += 1
        
        if i == 0:
            start_prob[tag] += 1
        else:
            transition[prev_tag][tag] += 1
        prev_tag = tag

def normalize(d):
    total = sum(d.values())
    return {k: v / total for k, v in d.items()}

start_prob = normalize(start_prob)
for tag in emission:
    emission[tag] = normalize(emission[tag])
    print(f"Emission probabilities for tag '{tag}': {emission[tag]}")
for prev in transition:
    transition[prev] = normalize(transition[prev])
    print(f"Transition probabilities from tag '{prev}': {transition[prev]}")
````

The Viterbi algorithm is commonly used for decoding the most likely sequence of tags given a sequence of words, based on the transition and observation probabilities estimated from the training data.

````python
def viterbi(sentence, states, start_p, trans_p, emit_p):
    V = [{}]
    path = {}

    for state in states:
        V[0][state] = start_p.get(state, 0) * emit_p[state].get(sentence[0], 1e-6)
        path[state] = [state]

    for t in range(1, len(sentence)):
        V.append({})
        new_path = {}

        for curr_state in states:
            max_prob, prev_state = max(
                (V[t - 1][y0] * trans_p[y0].get(curr_state, 1e-6) * emit_p[curr_state].get(sentence[t], 1e-6), y0)
                for y0 in states
            )
            V[t][curr_state] = max_prob
            new_path[curr_state] = path[prev_state] + [curr_state]

        path = new_path

    n = len(sentence) - 1
    prob, final_state = max((V[n][y], y) for y in states)
    return path[final_state]
````

In the main function, we would use a testing sentence and apply the Viterbi algorithm to find the most likely sequence of tags for that sentence. The output would be the tagged sentence, where each word is associated with its predicted part of speech tag.

````python
def main():
    states = list(tag_counts.keys())
    test_sentence = ["I", "like", "fish"]
    result = viterbi(test_sentence, states, start_prob, transition, emission)
    print("Sentence:", test_sentence)
    print("Predicted tags:", result)
    # error rate and f1 score can be calculated here by comparing with the actual tags
    actual_tags = ["PRON", "VERB", "NOUN"]  # Assuming we know the actual tags for evaluation
    error_rate = sum(1 for r, a in zip(result, actual_tags) if r != a) / len(actual_tags)
    print("Error Rate:", error_rate)
    print("F1 Score:", 1 - error_rate)  # Simplified F1 score for demonstration


if __name__ == "__main__":
    main()
````

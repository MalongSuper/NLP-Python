### Unicode with Python
Python 3 uses Unicode by default for its string type, which means you can work with a wide range of characters from different languages and symbol sets. Here are some key points about Unicode in Python:
1. **String Literals**: In Python 3, string literals are Unicode by default. You can simply use quotes to create a string, and it will be treated as a Unicode string.
2. **Encoding and Decoding**: When you need to convert between Unicode strings and byte strings (for example, when reading from or writing to files), you can use the `encode()` and `decode()` methods. The most common encoding is UTF-8.
3. **Character Representation**: Each character in a Unicode string can be represented by its code point, which is an integer value. You can use the `ord()` function to get the code point of a character and the `chr()` function to get the character from a code point.
4. **String Operations**: You can perform various operations on Unicode strings, such as concatenation, slicing, and formatting, just like with regular strings.
5. **Libraries**: There are many libraries available in Python for working with Unicode, such as `unicodedata` for accessing Unicode character properties and `regex` for advanced regular expression handling with Unicode support.
Overall, Python's support for Unicode makes it a powerful tool for handling text in multiple languages and character sets, making it ideal for natural language processing and other text-related tasks.

### Input texts
`text1 = 'Natural Language Processing'`

`text2 = 'Xử lý ngôn ngữ tự nhiên'`

One is in English and the other is in Vietnamese, which demonstrates the ability of Python to handle Unicode characters from different languages seamlessly.

```python
text1 = 'Natural Language Processing'
text2 = 'Xử lý ngôn ngữ tự nhiên'
```

### Decimal Representation
To see the decimal representation of each character in the input texts, we can use the `ord()` function, which returns the Unicode code point of a character. Here's how you can do it for both `text1` and `text2`:

```python
# Decimal Representation
print('Text 1\n', list(text1), "\n", [ord(char) for char in text1])
print('Text 2\n', list(text2), "\n", [ord(char) for char in text2])
```

### Hexadecimal Representation
In addition to the decimal representation, you can also view the hexadecimal representation of each character's Unicode code point using the `hex()` function. This can be particularly useful for understanding the encoding of characters in a more compact form. Here's how you can do it for both `text1` and `text2`:

```python
# Hexadecimal (Unicode Format)
print('Text 1\n', list(text1), "\n", [hex(ord(char)) for char in text1])
print('Text 2\n', list(text2), "\n", [hex(ord(char)) for char in text2])
```

### Binary Representation
You can also represent each character in binary format using the `bin()` function along with `ord()`. This will give you the binary representation of the Unicode code point for each character. Here's how you can do it for both `text1` and `text2`:

```python
# Binary Representation
print('Text 1\n', list(text1), "\n", [bin(ord(char)) for char in text1])
print('Text 2\n', list(text2), "\n", [bin(ord(char)) for char in text2])
```

### Encoding and Decoding
When working with Unicode strings, you may need to encode them into byte strings for storage or transmission, and decode byte strings back into Unicode strings when reading data. Here's how you can do it using UTF-8 encoding:

```python
# Encoding and Decoding
encoded1 = text1.encode("utf-8")
print(encoded1)
decoded1 = encoded1.decode("utf-8")
print(decoded1)

encoded2 = text2.encode("utf-8")
print(encoded2)
decoded2 = encoded2.decode("utf-8")
print(decoded2)
```

### Unicode Mapping
You can create a mapping of characters to their Unicode code points for both `text1` and `text2`. This can be done using a dictionary comprehension. Here's how you can do it:

```python
# Unicode Mapping
unicode_map1 = {char: ord(char) for char in text1}
print(unicode_map1)

unicode_map2 = {char: ord(char) for char in text2}
print(unicode_map2)
```

### Bytes Calculation
You can calculate the number of bytes required to encode each character in UTF-8. In UTF-8, characters can take 1 to 4 bytes depending on their code point. Here's how you can calculate the byte size for each character in both `text1` and `text2`:

```python
print('Text 1\n', list(text1), "\n", [len(char.encode("utf-8")) for char in text1], "\n", "Total bytes for Text 1:", sum(len(char.encode("utf-8")) for char in text1))
print('Text 2\n', list(text2), "\n", [len(char.encode("utf-8")) for char in text2], "\n", "Total bytes for Text 2:", sum(len(char.encode("utf-8")) for char in text2))
```
